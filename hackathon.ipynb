```python
import os
from langchain import PromptTemplate
from langchain_openai import ChatOpenAI  # Updated import
from langchain.chains import LLMChain
```

```python
load_dotenv()
```

```python
llm = ChatOpenAI(model_name="gpt-4")
```

```python
prompt_template = PromptTemplate(
    input_variables=["input_text"],
    template=(
        "You are an emoji generator. For the following text, generate a sequence of emojis that "
        "represent the words or sentiments expressed, without any labels or additional text. "
        "Return just the emojis in a single line, separated by spaces.\n\n"
        "Text: {input_text}\n"
        "Emojis:"
    ),
)
```

```python
emoji_chain = LLMChain(llm=llm, prompt=prompt_template)
```

```python
def convert_text_to_emoji(input_text):
    response = emoji_chain.invoke({"input_text": input_text})  # Use invoke method
    print("Response:", response)      
    return response['text'].strip()
```

```python
if __name__ == "__main__":
    input_text = "I'm eating pepporoni pizza"
    print("Input Text:", input_text)
    print("Emoji Output:", convert_text_to_emoji(input_text))
```
